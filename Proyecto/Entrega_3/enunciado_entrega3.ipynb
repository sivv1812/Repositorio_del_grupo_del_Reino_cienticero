### 📌 Análisis individual [3.0 puntos]

````python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import f1_score, classification_report, confusion_matrix

# ubicación de los datos y predicciones
DATA_DIR = '.\data'
PREDS_DIR = '.\predictions'
print(f"Data dir: {DATA_DIR}, Predictions dir: {PREDS_DIR}")

# Diccionario para guardar resultados y usarlos en la comparativa de predicciones
resultados_globales = {'batch': [], 'f1': [], 'precision': [], 'recall': []}

def evaluar_batch(batch_id):
    print(f"\n--- ANALIZANDO BATCH {batch_id} ---")
    
    #cargando los datos reales
    path_real = os.path.join(DATA_DIR, f'batch_t{batch_id}.parquet')
    if not os.path.exists(path_real):
        print(f" Error: No encuentro {path_real}")
        return None
    
    df_real = pd.read_parquet(path_real)
    col_target = 'items' if 'items' in df_real.columns else 'y_cat' 
    df_real['y_true'] = df_real[col_target].apply(lambda x: 1 if x > 0 else 0)

    # Cargando Predicción 
    path_pred = os.path.join(PREDS_DIR, f'predicciones_batch{batch_id}.csv')
    if not os.path.exists(path_pred):
        print(f" Error: No encuentro {path_pred}")
        return None

    df_pred = pd.read_csv(path_pred)
    df_pred['y_pred'] = 1 

    # cruzando los datos reales y predichos
    df_merged = df_real.merge(
        df_pred[['customer_id', 'product_id', 'y_pred']], 
        on=['customer_id', 'product_id'], 
        how='left'
    )
    df_merged['y_pred'] = df_merged['y_pred'].fillna(0).astype(int)
    
    #mostrando métricas
    report = classification_report(df_merged['y_true'], df_merged['y_pred'], output_dict=True)
    f1 = report['weighted avg']['f1-score']
    prec = report['weighted avg']['precision']
    rec = report['weighted avg']['recall']
    
    print(f"F1 Score (Weighted): {f1:.4f}")
    print(classification_report(df_merged['y_true'], df_merged['y_pred']))

    # guardando resultados para comparativa de predicciones
    resultados_globales['batch'].append(f'Batch {batch_id}')
    resultados_globales['f1'].append(f1)
    resultados_globales['precision'].append(prec)
    resultados_globales['recall'].append(rec)

    # creando matriz de confusión
    cm = confusion_matrix(df_merged['y_true'], df_merged['y_pred'])
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title(f'Matriz de Confusión - Batch {batch_id}')
    plt.xlabel('Predicción')
    plt.ylabel('Realidad')
    plt.show()
    
    return f1
````

#### 📌 Batch 1 (01/01/25 - 05/01/25) [0.75 puntos]

````python
# Ejecutando evaluar_batch para Batch 1
f1_b1 = evaluar_batch(1)
````

En este primer despliegue productivo, el modelo base obtuvo un F1-Score (Weighted) de 0.7036.

Al observar la Matriz de Confusión, notamos un comportamiento conservador con  Alta Precisión en la clase 1, Cuando el modelo predice compra, acierta la gran mayoría de las veces con una presición alrededor del 95\%. con respecto a los Falsos Negativos, el modelo está dejando pasar una cantidad considerable de clientes que sí compraron, clasificándolos como no compraron, siendo una posible perdida para la empresa. Esto establece nuestra línea base de desempeño. El modelo es seguro al presentar pocos falsos positivos pero le falta cobertura con un recall bajo.

#### 📌 Batch 2 (06/01/25 - 12/01/25) [0.75 puntos]

````python
# Ejecutando evaluar_batch para Batch 2
f1_b2 = evaluar_batch(2)
````

Durante la segunda semana, el desempeño se mantuvo prácticamente estable con un F1-Score de 0.7047.

La distribución de errores presentes en la matriz de confusión es muy similar a la semana anterior con unos Falsos Negativos altos de 2677. Con respecto a la detección de drift, el desempeño no mejoró de forma orgánica y los patrones de compra pueden haber cambiado de forma ligera, es en este punto donde nuestro componente detect_drift.py es crítico. Si las distribuciones de las variables clave como num_visit_per_week mostraron desviación con un PSI > 0.2, el sistema marcó la necesidad de un reentrenamiento para la siguiente semana de compras.

#### 📌 Batch 03 (13/01/25 - 19/01/25) [0.75 puntos]

````python
# Ejecutando evaluar_batch para Batch 3
f1_b3 = evaluar_batch(3)
````

Observamos un mejora en el rendimiento del modelo, alcanzando un F1-Score de 0.7566 mejor que la semana anterior.

Interpretando al  Pipeline, este aumento valida la eficacia de nuestro flujo automatizado en Airflow. Es altamente probable que, tras detectar el drift en la semana, el sistema ejecutó la rama de Reentrenamiento obteniendo un nuevo modelo de XGBoost. El modelo actualizado aprendió del nuevo comportamiento de los clientes, lo que da como resultado un  modelo que logró capturar mejor a los compradores reales, mejorando el equilibrio entre Precisión y Recall.

#### 📌 Batch 04 (20/01/25 - 26/01/25) [0.75 puntos]

````python
# Ejecutando evaluar_batch para Batch 4
f1_b4 = evaluar_batch(4)
````

El modelo demuestra robustez al cerrar el ciclo con un F1-Score de 0.7632, el más alto de todo el periodo.

La consistencia entre el Batch 3 y el Batch 4 indica que el reentrenamiento no fue una mejora unicamente para la semana 3 , sino que el modelo logró generalizar correctamente la nueva tendencia de los datos. La matriz de confusión muestra una reducción proporcional de los Falsos Negativos respecto al volumen total, confirmando que el modelo se adapto a la realidad actual de los clientes del negocio.

### 📌 Análisis comparativo [2.0 puntos]

````python
# Creando DataFrame con los resultados acumulados para comparación
df_res = pd.DataFrame(resultados_globales)

print("Tabla Resumen de Métricas:")
display(df_res)

# Gráfico de Tendencia para comparar batches
plt.figure(figsize=(10, 5))
plt.plot(df_res['batch'], df_res['f1'], marker='o', linestyle='-', linewidth=2, color='royalblue', label='F1 Score')
plt.plot(df_res['batch'], df_res['precision'], marker='s', linestyle='--', color='green', alpha=0.6, label='Precision')
plt.plot(df_res['batch'], df_res['recall'], marker='^', linestyle='--', color='orange', alpha=0.6, label='Recall')

plt.title("Evolución del Performance del Modelo (Batch 1 a 4)", fontsize=14)
plt.ylabel("Score")
plt.grid(True, alpha=0.3)
plt.legend()
plt.ylim(0, 1.0) 
plt.show()
````

Al analizar la evolución del F1-Score a través de las 4 semanas (Gráfico de Tendencia), identificamos dos fases claras:

1. Fase de Estancamiento (semana 1 y 2): El modelo original operó con métricas aceptables pero estáticas. Aquí el modelo presenta "Data Drift" al perder la capacidad de capturar la variabilidad de los nuevos datos presentados.

2. Fase de Adaptación (semana 2, 3 y 4):  la mayor mejora en las métricas fue en la semana 3 ,lo que demuestra el valor del MLOps, el pipeline al ser automatico permitió detectar el desvío y actualizar los pesos del modelo XGBoost sin una intervención manual.

En conlusión podemos decir que la implementación de un pipeline con Detección de Drift y Reentrenamiento Automático fue la decisión técnica relevante, permitiendo pasar de un modelo estático que se degradaba o estancaba, a un sistema dinámico, mejorando el F1-Score en más de un 8% (de 0.70 a 0.76) a lo largo del mes de enero, maximizando el valor de f1 score para SodAI Drinks.

### 📌 Conclusiones y Aprendizajes [1.0 puntos]

Con lo realizado en el proyecto se puede obtener estas conlusiones y aprendizajes:

### 1. Evolución del Modelo y Efecto del Reentrenamiento
La métrica más interesante de este proyecto fue el salto en el **F1-Score del 70% al 76%** observado entre el Batch 2 y el Batch 3.
Se diagnostico que en las primeras dos semanas, el modelo mostró un desempeño estancado (~0.70),, lo que sugiere que el modelo base entrenado con los datos originales estaba sufriendo de Data Drift; es decir, los patrones de compra de los clientes en Enero cambiaron respecto al histórico y el modelo antiguo no capturaba este comportamiento.
Al implementar Corrección Automática El pipeline detectó este cambio  y activó la rama de reentrenamiento. El resultado en el modelo generado por el reentrenamiento logró adaptarse a la nueva distribución de datos, recuperando 6 puntos porcentuales de rendimiento y manteniendo esa estabilidad en el Batch 4. **Esto valida la hipótesis central de MLOps: un modelo dinámico es superior a uno estático.**

### 2. Decisiones Técnicas de Alto Impacto
La decisión técnica más influyente no fue la elección del modelo de arbol XGBoost, sino la **arquitectura del Pipeline en Airflow con Branching**. Al implementar un nodo de decisión (`BranchPythonOperator`) basado en métricas estadísticas Test KS y PSI permitió automatizar el ciclo de vida del modelo. Sin esta mejora, el modelo estaria degradándose o estancándose en 0.70, lo que se traduciría en malas predicciones de ventas para SodAI, en otras palabras, perdidas para esta.

### 3. Hiperparámetros y Variables
Gracias al uso de **Optuna** integrado con **MLflow**, se obserba al ajuste de hiperparámetros no es trivial. Para poder adaptarse a los nuevos batches, parámetros como el `learning_rate` y `max_depth` fueron cruciales para matener sobreajuste controlado en periodos de tiempo cortos siendo en este caso semanales. Viendo la interpretabilidad del modelo podemos decir que el nivel de negocio, variables comportamentales como `num_visit_per_week` y `num_deliver_per_week` demostraron ser variables más fuertes que las características estáticas del cliente. Esto nos enseña que la frecuencia con la que visita el cliente va es mas util que caracteristicas de este para la predicción de las próximas compras.

### 4. Aprendizajes del Negocio (SodAI Drinks)
Podemos ver aprendizajes del negocio con este modelo al ver que el comportamiento de compra de bebidas es altamente estacional y sensible a tendencias de corto plazo. Un modelo que no se actualiza semanalmente pierde capacidad predictiva.  Al mejorar el Recall en los últimos batches al reducir falsos negativos, el negocio deja de perder oportunidades de venta. Es mejor sugerir productos a un cliente que quizás no compre que ignorar a un cliente que si va a  comprar.

### 5. Limitaciones Detectadas
Se presentarion limitaciones al ser un pipeline semanal (`@daily` o semanal), existe una ventana de tiempo "ciega" donde el modelo no se entera de cambios bruscos hasta que termina la semana. Tambien se presento limitaciones con los Clientes nuevos que aparecieron en los Batches 3 y 4 pueden tener predicciones menos precisas debido a la falta de historial de transacciones (`tx_raw`) profundo para generar los lags y features agregadas.
